```python
def saludar(nombre, sexo):
    sexo = sexo.lower()
    if sexo == "mujer":
        print(f"Bienvenida, {nombre}!")
    elif sexo == "hombre":
        print(f"Bienvenido, {nombre}!")
    else: 
        print(f"Bienvenid@, {nombre}!")
        
saludar("Ana", "mujer")
saludar("Carlos", "hombre")
saludar("Alex", "otro")       
    
```

```python
# Operador args

def sumar(*numeros):
    total = 0
    for numero in numeros:
        total += numero
    return total

resultado = sumar(1, 2, 3, 4, 5)
print(f"La suma es: {resultado}")
```
