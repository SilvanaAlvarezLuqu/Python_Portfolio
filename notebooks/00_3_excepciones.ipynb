# Excepciones

```python
def sumar_dos():
    while True:
        # pedir numeros
        a = input("Ingrese el primer número: ")
        b = input("Ingrese el segundo número: ")
        # intento convertirlo a entero y sumarlo
        try:
            resultado = int(a) + int(b)
        # si lanza una excepción, aviso y vuelvo a pedir los números
        except Exception as e: 
            print("Error: Por favor, ingrese solo números enteros.")
            print(f"ERROR {e}")
        # Si todo va bien, salgo del bucle
        else:
            break  
        # El finally se ejecuta siempre  
        finally:
            print("Intento de suma finalizado.")
    return resultado

print("La suma es:", sumar_dos())
```
