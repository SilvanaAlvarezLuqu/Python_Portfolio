```python
import tensorflow as tf
# Keras is bilt on top of TensorFlow
from tensorflow import keras

import matplotlib.pyplot as plt
%matplotlib inline
import numpy as np
```

```python
# Load the MNIST dataset
(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()
```

```python
print(f"Train set has {len(X_train)} samples")
print(f"Test set has {len(X_test)} samples")
```

```python
# Tamaño de cada imagen (28x28 píxeles)
X_train[0].shape
```

```python
# Matriz de pixeles de la primera imagen. Valores entre 0 y 255
X_train[0]
```

```python
plt.matshow(X_train[0])
```

```python
y_train[0]
```

```python
# Scale datasets to range 0-1
X_train = X_train / 255
X_test = X_test / 255
```

```python
# Flatten x_train
X_train_flat = X_train.reshape(len(X_train), 28*28)
X_test_flat = X_test.reshape(len(X_test), 28*28)

print(X_train_flat.shape)
print(X_test_flat.shape)
```

```python
X_train_flat[0]
```

# 1) Simple Neuronal Network
Solo tendrá 2 capas. La capa de entrada con 784 neuronas (28x28 pixeles) y una capa de salida con 10 neuronas (0-9 dígitos).

### Entrenar y evaluar el modelo

```python
# Sequential significa que las capas se apilan una tras otra
# Acepta cada capa como un elemento
# Dense: TODAS las neuronas de una capa están conectadas a TODAS las neuronas de la siguiente capa

# Definir el modelo
model = keras.Sequential([
    keras.layers.Dense(10, input_shape=(784,), activation="sigmoid")
])

# Definir el optimizador, la función de pérdida y las métricas
model.compile(optimizer="adam",
              loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])

# Entrenar el modelo (train set)
model.fit(X_train_flat, y_train, epochs=5)
```

```python
# Evaluar el modelo (test set)
model.evaluate(X_test_flat, y_test)
```

### Algunas predicciones

```python
plt.matshow(X_test[2])
```

```python
y_pred = model.predict(X_test_flat)
y_pred[2]
```

```python
np.argmax(y_pred[2])
```

### Confusion Matrix

```python
# Take the index with the highest probability
y_predicted_labels = [np.argmax(i) for i in y_pred]

# Compute the confusion matrix
cm = tf.math.confusion_matrix(labels = y_test, predictions=y_predicted_labels)
cm
```

```python
import seaborn as sn

plt.figure(figsize=(10,7))
sn.heatmap(cm, annot=True, fmt='d')
plt.xlabel('Predicted')
plt.ylabel('Truth')
```

# 2) Using Hidden Layers

```python
# Definir el modelo
# El 100 es el número de neuronas en la capa oculta (menor al número de entradas pero elegido random)

model = keras.Sequential([
    keras.layers.Dense(100, input_shape=(784,), activation="relu"), # Only first layers needs input_shape
    keras.layers.Dense(10, activation="sigmoid")
])

# Definir el optimizador, la función de pérdida y las métricas
model.compile(optimizer="adam",
              loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])

model.fit(X_train_flat, y_train, epochs=5)
```

```python
# Evaluar el modelo (test set)
model.evaluate(X_test_flat, y_test) # con una capa oculta mejora el accuracy de 92 a 97
```

```python
y_predicted = model.predict(X_test_flat)
y_predicted_labels = [np.argmax(i) for i in y_predicted]
cm = tf.math.confusion_matrix(labels=y_test,predictions=y_predicted_labels)

plt.figure(figsize = (10,7))
sn.heatmap(cm, annot=True, fmt='d')
plt.xlabel('Predicted')
plt.ylabel('Truth')
```

# 3) Flatten Layer usage

```python
# Definir el modelo
# El 100 es el número de neuronas en la capa oculta (menor al número de entradas pero elegido random)

model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28,28)), # Nueva capa Flatten
    keras.layers.Dense(100, activation="relu"), # Only first layers needs input_shape
    keras.layers.Dense(10, activation="sigmoid")
])

# Definir el optimizador, la función de pérdida y las métricas
model.compile(optimizer="adam",
              loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])

model.fit(X_train, y_train, epochs=5)
```

## Tensorboard

```python
# Definir el modelo
# El 100 es el número de neuronas en la capa oculta (menor al número de entradas pero elegido random)

model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28,28)), # Nueva capa Flatten
    keras.layers.Dense(100, activation="relu"), # Only first layers needs input_shape
    keras.layers.Dense(10, activation="sigmoid")
])

tb_callback = keras.callbacks.TensorBoard(log_dir="logs/", histogram_freq=1)

# Definir el optimizador, la función de pérdida y las métricas
model.compile(optimizer="SGD",
              loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])

model.fit(X_train, y_train, epochs=5, callbacks=[tb_callback])
```
