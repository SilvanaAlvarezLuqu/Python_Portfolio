```python
import math
```

# Sigmoid
- Convierte cualquier real en valores entre 0 y 1.
- Función suave. Fácil selección de prob máxima para clasificación.

```python
def sigmoid(x):
    return 1 / (1 + math.exp(-x))
```

```python
print(sigmoid(100))
print(sigmoid(1))
print(sigmoid(-56)) # very close to 0   
print(sigmoid(0.5))
```

# Tanh
- Función suave que convierte cualquier real en valores entre -1 y 1.
- Centrada en 0, lo que puede ayudar en la convergencia del modelo. Recomendada para capas ocultas.

```python
def tanh(x):
    return (math.exp(x) - math.exp(-x)) / (math.exp(x) + math.exp(-x))
```

```python
print(tanh(100))
print(tanh(1))
print(tanh(-56)) # very close to 0   
print(tanh(0.5))
```

Con las funciones sigmoid y tanh podemos caer en vanishing gradient por derivadas cercanas a 0 en los valores extremos.

# ReLU
- Función no lineal que convierte valores negativos en 0 y mantiene positivos.
- Computacionalmente eficiente y ayuda a mitigar el problema de vanishing gradient.
- Estándar si no se sabe qué función usar.

```python
def relu(x):
    return max(0,x)
```

```python
print(relu(100))
print(relu(1))
print(relu(-56)) # very close to 0   
print(relu(0.5))
```

# Leaky ReLU

- Variante de ReLU que permite un pequeño gradiente para valores negativos.
- Ayuda a evitar neuronas muertas y mejora el aprendizaje en algunas situaciones.

```python
def leakyrelu(x, alpha=0.01):
    return max(alpha * x, x)
```

```python
print(leakyrelu(100))
print(leakyrelu(1))
print(leakyrelu(-56)) # very close to 0   
print(leakyrelu(0.5))
```
