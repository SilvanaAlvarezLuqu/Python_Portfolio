```python
ingreso_mensual = 8000
gastos_mensuales = 5000

if ingreso_mensual > 10000:
    if ingreso_mensual - gastos_mensuales < 0:
        print("Estás en deficit.")
    elif ingreso_mensual - gastos_mensuales >3000:
        print("Tienes un buen margen de ahorro.")
    else:
        print("Gastas casi todo lo que ganas.")
elif ingreso_mensual >= 5000:
    print("Tienes un ingreso medio.")
elif ingreso_mensual > 1000:
    print("Tienes un ingreso bajo.")
else :
    print("Eres pobre.")
```

## Métodos de listas
- len(): Devuelve el número de elementos en la lista.
- append(): Añade un elemento al final de la lista.
- insert(): Inserta un elemento en una posición específica.
- extend(): Añade todos los elementos de una lista (u otro iterable) al final de la lista actual.
- remove(): Elimina la primera aparición de un elemento específico.
- pop(): Elimina y devuelve el último elemento de la lista (o un elemento en unaposición específica).
- sort(): Ordena los elementos de la lista en orden ascendente o según un criterio personalizado.
- reverse(): Invierte el orden de los elementos en la lista.
- index(): Devuelve el índice de la primera aparición de un elemento específico.


## Métodos de diccionarios
- keys(): Devuelve una vista de las claves del diccionario.
- values(): Devuelve una vista de los valores del diccionario.
- items(): Devuelve una vista de los pares clave-valor del diccionario.
- get(): Devuelve el valor asociado a una clave específica, o un valor predeterminado si la clave no existe.
- pop(): Elimina una clave específica y devuelve su valor.
- clear(): Elimina todos los elementos del diccionario.

```python
otros_cursos_min = 2.5
otros_cursos_max = 7
otros_cursos_mean= 4
dalto_curso = 1.5

dif_min = 100 - dalto_curso / otros_cursos_min*100
dif_max = 100 - dalto_curso / otros_cursos_max*100
dif_mean = 100 - dalto_curso / otros_cursos_mean*100

print(f"El curso de Dalto es un {dif_min:.2f}% más rápido que el curso más rápido.")
print(f"El curso de Dalto es un {dif_max:.2f}% más rápido que el curso más lento.")
print(f"El curso de Dalto es un {dif_mean:.2f}% más rápido que el curso promedio.")

# Mostrar diferencias si los cursos durarán 10 horas
print(f"Ver 10 horas del curso de dalto equivale a ver {otros_cursos_mean/dalto_curso*10:.2f} horas del curso promedio.")
print(f"Ver 10 horas de otros cursos equivale a ver {dalto_curso/otros_cursos_mean*10:.2f} horas del curso de dalto.")
```

## BUCLES

```python
animales = ["gato", "perro", "loro", "cocodrilo", "pez"]
numeros = [10,35,23,47,59]

for animal in animales:
    print(f"El animal es: {animal}")
    
for numero,animal in zip(numeros,animales):
    print(f"Recorriendo la lista 1: {numero}")
    print(f"Recorriendo la lista 2: {animal}")    
    
for num in range(7):
    print(num)       
    
# Recorrer una lista con índice
for num in enumerate(numeros):
    indice = num[0]
    valor = num[1]
    print(f"El índice es: {indice}, el valor es: {valor}")    
```

```python
for numero in numeros:
    print(f"El número es: {numero}")
else:
    print("Se terminó de recorrer la lista.")    # Se muestra 1 sola vez al finalizar el bucle
```

```python
diccionario = {"Nombre": "Silvana",
               "Apellido": "Alvarez",
               "Edad": 25}

for datos in diccionario.items():
    print(datos)
    key = datos[0]
    value = datos[1]
    print(f"La clave es: {key}, el valor es: {value}")
```

```python
frutas = ["manzana", "banana", "cereza", "durazno"]

for fruta in frutas:
    if fruta == "cereza":
        continue    # Salta a la siguiente iteración
    print(f"Me voy a comer un(a) {fruta}")
    if fruta == "banana":
        break       # Sale del bucle
print("Fin del bucle")
```
