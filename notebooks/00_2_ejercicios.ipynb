```python
# Función para obtener el profesor y el asistente según la edad
def obtener_compañeros(cantidad_compañeros):
    
    # Crear lista de compañeros
    compañeros = []
    
    # Bucle para pedir la información de cada compañero
    for i in range(cantidad_compañeros):
        nombre = input("Ingrese el nombre del compañero")
        edad = int(input("Ingrese la edad del compañero"))
        compañero = (nombre, edad)
        # Agregar información a la lista
        compañeros.append(compañero)
        
    # Ordenar de menor a mayor por edad    
    compañeros.sort(key= lambda x: x[1])
    
    # Acceder al nombre del asistente y del profesor
    asistente = compañeros[0][0]
    profesor = compañeros[-1][0]
    
    # Retornar una tupla
    return asistente, profesor    

# Desempaquetar la tupla retornada
asistente, profesor = obtener_compañeros(5)

# Mostrar resultados
print(f"El profesor es {profesor} y el asistente es {asistente}")
```

```python
# Mi solución
def numeros_primos(num):
    primos= []
    for num in range(2, num):
        divisores = [i for i in range(1, num+1) if num % i == 0]
        if len(divisores) == 2:
            primos.append(num)
    return primos
    
primos = numeros_primos(13)
print(primos)     

# Solución del curso
def es_primo(num):
    for i in range(2, num-1):
        if num % i == 0: return False
    return True

def primos_hasta(num):
    primos = []
    for i in range(2, num):
        resultado = es_primo(i)
        if resultado == True: primos.append(i)
    return primos
    
resultado = primos_hasta(13)
print(resultado)
```

```python
# Serie de Fibonacci en un rango dado
def fibonacci(num):
    a,b = 0,1
    serie = [0]
    for i in range(num):
        if b > num: return serie
        else:
            serie.append(b)
            a,b = b, a+b
            
resultado = fibonacci(33)
print(resultado)            
            
        
```
